```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 1. Load data
data = pd.read_csv("./Data/Salary_Data.csv")
data.head()
```

```python
X = data["YearsExperience"].values
salary = data["Salary"].values
print(X)
print(salary)
```

```python
# 2. Convert salary to binary classes
threshold = 50000
y = np.where(salary >= threshold, 1, 0)

print("\nYearsExperience (first 5):", X[:10])
print("\nConverted Salary to classes (0=Low, 1=High):")
print(y[:10])
```

```python
# 3. Define Sigmoid Function
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

print("\nSigmoid test for z=0:", sigmoid(0))
```

```python
# 4. Initialize Parameters
m = 0.0      # weight (how much y is dependent in the years of experiene)   
c = 0.0      # bias (starting/ initial value )
lr = 0.01    # learning rate
epochs = 100
n = len(y)
''' BOTH m and c here is zero because it is just a initial point the model will gradually learn from this'''

print(f"\nInitial weight (m): {m}")
print(f"Initial bias (c): {c}")
```

```python
# 5. Training Loop (Gradient Descent)
for i in range(epochs):
    z = m * X + c
    y_pred = sigmoid(z)

    # Gradients
    dm = (1/n) * np.sum(X * (y_pred - y))  # gradient for weight  (dm → how m should change)

    ## dm = average error influenced by X
    # How much weight m is causing error.
    '''
    (y_pred - y) → error
    X * error → how input affects error
    sum → total effect
    1/n → average
    '''
    
    
    dc = (1/n) * np.sum(y_pred - y)      # Gradient for bias c    (dc → how c should change)
    '''
    dc = average error
    How much bias c is causing error.”
    '''
    # Update parameters 
    # (NEW VALUES = OLD VALUES - CORRECTION) where correction=learning rate x gradient)
    m = m - lr * dm
    c = c - lr * dc

    # Log loss every 200 epochs
    if i % (epoch;s//10) == 0:
        loss = (-1/n) * np.sum(
            y * np.log(y_pred + 1e-9) +
            (1 - y) * np.log(1 - y_pred + 1e-9)
        )
        print(f"\nEpoch {i}")
        print(f"Loss: {loss}")
        print(f"Weight (m): {m}")
        print(f"Bias (c): {c}")
```

```python
# 6. Final Evaluation
z_final = m * X + c
probabilities = sigmoid(z_final)
print(probabilities)

print("\nSigmoid probabilities (first 10):")
print(probabilities[:10])

y_pred_class = np.where(probabilities >= 0.5, 1, 0)

print("\nPredicted Classes (first 10):")
print(y_pred_class[:10])

accuracy = np.mean(y_pred_class == y)
print(f"\nModel Accuracy: {accuracy}")
```
